# Learning APIs - Google Books

Google Books Documentation - https://developers.google.com/books/docs/overview

## Importing Required Libraries

```python
import requests
import json
```

## Building the Endpoint and Parameters for Google Books

```python
isbn = "9780345806789" 
base_url = "https://www.googleapis.com/books/v1/volumes"
payload = { # data that's being sent and recieved
    "q": f"isbn:{isbn}",
    "maxResults": 1
}
print(f"Target URL: {base_url}")
print(f"Payload: {payload}")
```

## Checking the Server Connection

```python
response = requests.get(base_url, params=payload) # combining the base_url and payload to send
print(f"Status Code: {response} {response.reason}") # verifying the server response code
print(f"Generated URL: {response.url}")
```

## Inspecting the Raw Data

```python
data = response.json() # the raw data
print("Root Keys:", data.keys()) # output the keys of the main dictionary
print(json.dumps(data, indent=2))
```

## Extracting Specific Information

```python
items_list = data.get("items", [])
if items_list:
    first_book = items_list[0]
    volume_info = first_book.get("volumeInfo", {})
    title = volume_info.get("title", "Unknown")
    authors = volume_info.get("authors", [])
    publisher = volume_info.get("publisher", "Unknown")
    publishedDate = volume_info.get("publishedDate", [])
    page_count = volume_info.get("pageCount", 0)
    print_type = volume_info.get("printType", "Unknown")
    formatted_print_type = print_type.title()
    categories = volume_info.get("categories", [])
    description = volume_info.get("description", "Unknown")
    print(f"Title: {title}")
    print(f"Author: {authors}")
    print(f"Publisher: {publisher}")
    print(f"Published Date: {publishedDate}")
    print(f"Page Count: {page_count}")
    print(f"Print Type: {formatted_print_type}")
    print(f"Genres: {categories}")
    print(f"Description: {description}")
```
